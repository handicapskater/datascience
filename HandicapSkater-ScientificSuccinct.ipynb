# Skating mobility evidence: compact review

**One person, repeated observations, published aggregate evidence.** The central functional result is substantially less distance during the walking challenge and greater distance with PT skating. Walking also shows higher vertical acceleration and jerk than Mall skating.

The original WHOOP/Strava notebook established the longitudinal record and hypotheses that informed the legal notebook design. Those notebooks already included targeted Kubios/Polar H10 evidence; the current work refines the combined longitudinal-plus-episodic approach with more specific metrics and validation. This is scientific continuity, not a replacement or retrospective rewriting of the legal record. See the [research evolution companion](docs/research_evolution.md).

## Read this result with its limits

- Function and mechanics converge on 43/43 eligible Pareto dates; adding shock-spike rate yields 38/43.
- Longitudinal skating covers 598 activities / 589 dates, median 11.47 miles, across 2020–2026.
- PT restores distance without uniformly returning mechanics to Mall. The MAD-based impact count runs opposite to shock-spike count.
- Fixed order, duration and environment constrain causal interpretation. Telemetry does not identify pain or internal joint force.
- FSI/CSS summarize their governed inputs; they do not independently prove a need for accommodation. No new score or formula is introduced.

This supports a functional-accommodation argument alongside the clinical record. See [Public](HandicapSkater-Public.ipynb) for the narrative and [Scientific](HandicapSkater-Scientific.ipynb) for methods, scope and provenance.
